## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

### Ranked Queue

This baseline ranks pages using a simple CTR optimization rule. Pages with high impressions and low CTR receive higher scores because they already have visibility but are missing potential clicks.

The output is written to:

`work/outputs/baseline_action_score.csv`

```python
import pandas as pd
import os

# Take a sample for the baseline
sample = dataset["train"][:5000]

df = pd.DataFrame(sample)

# Calculate CTR
df["ctr"] = df.apply(
    lambda x: x["gsc_clicks"] / x["gsc_impressions"]
    if x["gsc_impressions"] > 0 else 0,
    axis=1
)

# Simple baseline score:
# High impressions + Low CTR = Higher score
df["baseline_score"] = df["gsc_impressions"] * (1 - df["ctr"])

# Reason code
df["reason_code"] = "LOW_CTR_HIGH_IMPRESSIONS"

# Action label
df["action"] = "CTR_OPTIMIZATION"

# Rank pages
df = df.sort_values("baseline_score", ascending=False)

# Create output folder if needed
os.makedirs("../outputs", exist_ok=True)

# Save CSV
output_file = "../outputs/baseline_action_score.csv"
df.to_csv(output_file, index=False)

print("Saved:", output_file)

# Show top 10
df.head(10)
```

Pages are prioritized because they have high search impressions but relatively low click-through rates. These pages are likely to benefit from title, meta description, or content improvements.

The review below explains why each page was selected and what could make the recommendation incorrect.
