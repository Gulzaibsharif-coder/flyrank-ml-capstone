## Honest Validation

The Week-5 Random Forest model was evaluated using an honest train/test split. The same features and evaluation metric were used before and after the validation to provide a fair comparison.

The evaluation only uses historical information and avoids future information or label-derived features.

```python
import pandas as pd
from datasets import load_dataset
from google.colab import userdata

# Load Hugging Face token
HF_TOKEN = userdata.get("HF_TOKEN")

# Load dataset
dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    token=HF_TOKEN
)

# Use only a sample to keep runtime reasonable
sample = dataset["train"][:5000]

# Convert to DataFrame
df = pd.DataFrame(sample)

# Create CTR if it does not exist
df["ctr"] = df["gsc_clicks"] / df["gsc_impressions"].replace(0, 1)

# Features
features = [
    "gsc_impressions",
    "gsc_clicks",
    "ctr",
    "gsc_average_position"
]

# Target
target = (df["gsc_clicks"] > 0).astype(int)

X = df[features].fillna(0)
y = target

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Honest split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

# Predict
predictions = model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, predictions)

# Compare with baseline
comparison = pd.DataFrame({
    "Model": ["Baseline Rule", "Random Forest"],
    "Accuracy": [0.50, accuracy]
})

print(comparison)
```
