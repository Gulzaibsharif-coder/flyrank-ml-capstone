## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### Data Limits

- The dataset contains anonymized client and content identifiers.
- The analysis is limited to the reporting period available in the FlyRank warehouse.
- Some performance metrics depend on Google Search Console (GSC) and Google Analytics 4 (GA4) availability.
- Results should not be generalized beyond the available observation period.

```python
sample = dataset["train"][:10000]

dates = sample["report_date"]

print("Earliest report date in sample:", min(dates))
print("Latest report date in sample:", max(dates))
print("Unique reporting dates in sample:", len(set(dates)))
```
